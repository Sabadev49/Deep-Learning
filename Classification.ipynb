```python
import tensorflow as tf
import numpy as np
from tensorflow.keras.layers import Input, Dense
from tensorflow.keras.models import Sequential
```

```python
X_cls = np.array([
    [2, 0.50],
    [3, 0.60],
    [4, 0.65],
    [1, 0.40],
    [5, 0.75],
    [6, 0.80],
    [7, 0.85]
], dtype=float)

y_cls = np.array([0, 0, 0, 0, 1, 1, 1], dtype=float)
```

```python
model=Sequential()
model.add(Input(shape=(2,)))
model.add(Dense(units=1, activation='sigmoid'))
```

```python
X_cls.shape
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
history=model.fit(X_cls,y_cls,epochs=2000,batch_size=4)
```

```python
model.evaluate(X_cls,y_cls)
```

```python
pred1=model.predict(np.array([[2,0.55]]))
pred2=model.predict(np.array([[4.5, 0.70]]))
print(pred1,pred2)
#1: rad 2:ghabool
```

```python
final_accuracy=history.history['accuracy'][-1]
print(final_accuracy)
```
